````python
#Prerequisites
# uv add langchain_openai
# uv add langchain
# uv add langchain_community
# uv add pypdf

import os
from langchain_openai import ChatOpenAI #
from langchain_community.document_loaders import PyPDFLoader # Loads the PDF file
from langchain_text_splitters import RecursiveCharacterTextSplitter # Splits the text into chunks
from dotenv import load_dotenv # Loads the environment variables


load_dotenv()
````

````python
text_data = PyPDFLoader("NovaS.pdf").load()
text_data
for page in text_data:
    page.metadata['source'] = 'NovaS.pdf'
````

````python
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)

chunks = splitter.split_documents(text_data)
chunks
````

````python
from langchain_openai import OpenAIEmbeddings
embed_model = OpenAIEmbeddings(model = 'text-embedding-3-small')

embedded_chunks = embed_model.embed_documents([chunk.page_content for chunk in chunks])
embedded_chunks[0]
````

## **Step 2 and 3: Create Embeddings and Store in Vector DB**

````python
from langchain_community.vectorstores import Chroma

Chroma.from_documents(
    documents = chunks,
    embedding = embed_model,
    persist_directory = './vector_db'
)
````

## **Step 4: Connection to Vector DB and Retrieval**

````python
chroma_db_conn = Chroma(
    embedding_function = embed_model,
    persist_directory = './vector_db'
)
````

````python
chroma_db_conn.similarity_search('When NovaSphere Technologies was founded?', k=3)
````

````python
chroma_db_conn.similarity_search('By 2019, how many employees did NovaSphere Technologies have?', k=3)
````

## **Step 5: Pass context to LLM and recieve responses**

````python
# temperature is parameter to control randomness of the model, 0 is deterministic and 1 is random. Anything in between means the model will be somewhere in between.
llm = ChatOpenAI(model = 'gpt-3.5-turbo', temperature = 0) 
llm
````

````python
user_query = input("Enter your question:")
relevant_chunks = chroma_db_conn.similarity_search(user_query, k=10)

relevant_chunks_content = []
for i, chunk in enumerate(relevant_chunks):
    relevant_chunks_content.append(chunk.page_content)
relevant_chunks_content = str(relevant_chunks_content)

print(relevant_chunks_content)
llm.invoke(f"{user_query}, Use the following context to answer the question: {relevant_chunks_content}")
````

## Notes: Retrieval-Augmented Generation (RAG)

### What Is RAG?

RAG is a pattern that **augments an LLM's knowledge** by retrieving relevant documents from an external source and injecting them into the prompt as context. This solves two fundamental LLM limitations:
1. **Knowledge cutoff** — the model only knows what was in its training data.
2. **Hallucination** — without grounding, models may fabricate facts. RAG provides factual anchors.

### The RAG Pipeline (5 Steps)

```
                     INDEXING (offline, one-time)
┌──────────────┐    ┌──────────────┐    ┌──────────────┐    ┌──────────────┐
│  Load Docs   │ →  │  Split into  │ →  │   Embed      │ →  │  Store in    │
│  (PyPDFLoader)│    │  Chunks      │    │   Chunks     │    │  Vector DB   │
└──────────────┘    └──────────────┘    └──────────────┘    └──────────────┘

                     RETRIEVAL + GENERATION (online, per query)
┌──────────────┐    ┌──────────────┐    ┌──────────────┐    ┌──────────────┐
│  User Query  │ →  │  Embed Query │ →  │  Similarity  │ →  │  LLM + Context│
│              │    │              │    │  Search (k=N)│    │  → Answer     │
└──────────────┘    └──────────────┘    └──────────────┘    └──────────────┘
```

### Step-by-Step Breakdown

#### Step 1: Document Loading (`PyPDFLoader`)

- Reads the PDF page by page, extracting text and metadata (author, creation date, page number).
- Each page becomes a LangChain `Document` object with `.page_content` and `.metadata`.
- Other loaders exist for different formats: `CSVLoader`, `TextLoader`, `UnstructuredHTMLLoader`, etc.

#### Step 2: Text Splitting (`RecursiveCharacterTextSplitter`)

Why split? A single embedding cannot capture the nuance of a long document. Smaller chunks allow more precise retrieval.

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
```

- **`chunk_size=100`**: Target ~100 characters per chunk. In production, 500–1000 characters (or 100–500 tokens) is more common.
- **`chunk_overlap=20`**: 20 characters of overlap between consecutive chunks to avoid cutting sentences mid-thought.
- **Recursive strategy**: Tries to split on `\n\n` → `\n` → ` ` → `` (character) in order, preserving natural boundaries.

| Parameter | Too small | Too large |
|---|---|---|
| `chunk_size` | Loses context, fragments sentences | Single embedding covers too many topics, retrieval becomes imprecise |
| `chunk_overlap` | Hard cuts at chunk boundaries | Redundant storage and embedding cost |

#### Step 3: Embedding (`OpenAIEmbeddings`)

Each chunk is converted into a 1536-dimensional vector using `text-embedding-3-small`. These vectors capture semantic meaning — chunks about similar topics will have nearby vectors.

#### Step 4: Vector Store (`ChromaDB`)

- **ChromaDB** is a lightweight, open-source vector database that runs locally (no server needed).
- `Chroma.from_documents()` embeds all chunks and stores both the vectors and the original text/metadata.
- `persist_directory` saves the index to disk so you don't re-embed on every restart.
- **Similarity search** uses approximate nearest neighbors (ANN) to find the k most relevant chunks for a query.

Other vector stores: Pinecone (managed cloud), Weaviate, Qdrant, FAISS (Facebook, in-memory), pgvector (Postgres extension).

#### Step 5: Retrieval + Generation

```python
relevant_chunks = chroma_db_conn.similarity_search(user_query, k=10)
llm.invoke(f"{user_query}, Use the following context: {relevant_chunks_content}")
```

1. **Embed the query** using the same embedding model.
2. **Retrieve top-k chunks** by cosine similarity.
3. **Stuff** the retrieved chunks into the prompt as context.
4. **LLM generates** an answer grounded in the provided context.

### Retrieval Strategies

| Strategy | How it works | Pros | Cons |
|---|---|---|---|
| **Similarity search** (this notebook) | Cosine similarity between query and chunk embeddings | Simple, fast | No re-ranking, no diversity |
| **MMR (Maximal Marginal Relevance)** | Balances relevance with diversity | Avoids redundant chunks | Slightly slower |
| **Hybrid search** | Combines vector similarity + keyword (BM25) search | Best of both worlds | More complex setup |
| **Re-ranking** | Retrieve broadly, then re-rank with a cross-encoder | Higher precision | Extra model call |

### Key Observations from This Notebook

1. **Duplicate results**: The similarity search returns duplicate chunks (same content appears multiple times). This happens because `Chroma.from_documents()` was likely called multiple times, inserting duplicates. In production, deduplicate before insertion or use unique IDs.

2. **Naive prompting**: The context is stuffed directly into an f-string. Production systems use prompt templates with clear instructions like "Answer ONLY from the provided context. If the context doesn't contain the answer, say 'I don't know.'"

3. **No conversation memory**: Each query is independent. For multi-turn RAG, you'd maintain chat history and potentially reformulate queries.

4. **`temperature=0`**: Set for deterministic, factual responses — appropriate for RAG where you want the model to faithfully report what the documents say rather than get creative.

### Common RAG Pitfalls

- **Wrong chunk size**: Too small fragments meaning, too large dilutes relevance.
- **No metadata filtering**: Real systems filter by source, date, or category before similarity search.
- **Embedding model mismatch**: Query and documents must use the same embedding model.
- **Context window overflow**: Stuffing too many chunks can exceed the model's context limit or degrade quality (lost-in-the-middle effect).
- **No evaluation**: Without measuring retrieval recall and answer quality, you're flying blind.
