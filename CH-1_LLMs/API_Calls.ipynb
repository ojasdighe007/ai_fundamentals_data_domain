## **OpenAI API Calls using OpenAI Python SDK**

````python
from openai import OpenAI
client = OpenAI()
#Better to use Langchain for this, but this is just for testing
response = client.responses.create(
    model="gpt-5-mini-2025-08-07",
    input="Write a one-sentence bedtime story about a unicorn."
)

print(response.output_text)
````

## Notes: OpenAI API Calls

### The OpenAI Python SDK

The `openai` Python package is the official client for interacting with OpenAI's APIs. It handles authentication (via `OPENAI_API_KEY` environment variable), request formatting, retries, and response parsing.

```python
from openai import OpenAI
client = OpenAI()  # reads OPENAI_API_KEY from env
```

### Responses API vs Chat Completions API

OpenAI offers two main APIs for text generation:

| API | Method | Stateful? | Built-in Tools? | Use case |
|---|---|---|---|---|
| **Responses API** (`client.responses.create`) | Newer, recommended | Yes (can maintain conversation state server-side) | Yes — web search, file search, code interpreter | New applications, agentic workflows |
| **Chat Completions API** (`client.chat.completions.create`) | Legacy, widely adopted | No (stateless, you manage history) | No — you must bind tools yourself | Existing apps, fine-grained control |

This notebook uses the **Responses API**, which is OpenAI's newer interface. The key parameter is `input` (a string or message array) rather than `messages`.

### How an LLM Call Works End-to-End

```
Your Python code
    ↓
OpenAI SDK serializes request (model, input, parameters)
    ↓
HTTPS POST to api.openai.com/v1/responses
    ↓
OpenAI routes to the requested model (gpt-5-mini)
    ↓
Model runs inference (tokenize → transformer forward pass → sampling)
    ↓
Response streamed/returned as JSON
    ↓
SDK parses into Response object → response.output_text
```

### Key Parameters

- **`model`**: Which model to use. Model choice affects cost, speed, and quality. `gpt-5-mini` is optimized for speed and cost while maintaining strong performance.
- **`input`**: The prompt text or message array.
- **`temperature`** (not shown here, default 1.0): Controls randomness. 0 = deterministic, 1 = creative. Lower values for factual tasks, higher for creative tasks.
- **`max_output_tokens`**: Cap on response length.
- **`top_p`**: Nucleus sampling — alternative to temperature for controlling diversity.

### LLM (API) vs ChatGPT (Product)

A common misconception: calling the API is **not** the same as using ChatGPT.

- **API call** = raw model, no memory, no tools, no system prompt (unless you provide one). You get back a single completion.
- **ChatGPT** = model + agent loop + tools (web search, code interpreter, DALL-E) + memory + system prompt + UI. It's a full application built on top of the same models.

This notebook demonstrates the simplest possible interaction: one prompt in, one completion out, no conversation history.

### LangChain vs Raw SDK

The comment in the code mentions LangChain being "better for this." Why?

| Aspect | Raw OpenAI SDK | LangChain |
|---|---|---|
| **Simplicity** | Direct, minimal abstraction | More boilerplate for simple calls |
| **Tool calling** | Manual JSON schema management | `@tool` decorator + `bind_tools()` |
| **Agent loops** | Build your own | `create_agent()` out of the box |
| **Provider switching** | Rewrite code for each provider | Swap `ChatOpenAI` → `ChatAnthropic` etc. |
| **Chains/Pipelines** | Manual composition | LCEL (LangChain Expression Language) |

For a single API call, the raw SDK is simpler. For building agents, RAG pipelines, or multi-step chains, LangChain provides useful abstractions.
